# Cleaning

```python
import warnings
```

## Price

```python
def to_plausible_price(x):
    try:
        x = float(x)
    except:
        warnings.warn(f"cant turn this value to numeric: {x}")
        return -1
        
    if x <= 1000 or x >= 999999:
        return -1
    else:
        return x
```

```python
num_price = df['price'].dropna().apply(lambda x: to_plausible_price(x))
num_price = num_price[num_price > -1]
print(f"$: min: {num_price.min()} , avg: {num_price.mean()}, max: {num_price.max()}")
min_price, max_price = num_price.min(), num_price.max()
```

## Year

```python
import warnings
def to_plausible_year(x):
    try:
        x = int(x)
    except:
        warnings.warn(f"cant turn this value to numeric: {x}")
        return -1
        
    if x <= 1900 or x >= 2025:
        return -1
    else:
        return x
```

```python
num_year = df['year'].dropna().apply(lambda x: to_plausible_year(x))
num_year = num_year[num_year > -1]
print(f"year: min: {num_year.min()} , mode: {num_year.mode().squeeze()}, max: {num_year.max()}")
min_year, max_year = num_year.min(), num_year.max()
```

## Mileage

```python
import re
def to_plausible_mileage(x):
    k_pattern = r'\b(\d+(?:\.\d+)?)[kK]\b'
    if isinstance(x, str):
        if found_k := re.findall(k_pattern, x):
            found_k = found_k[0]
            converted = int(float(found_k.rstrip('kK'))*1000)
            print(f"Converting {found_k}k -> {converted}")
            return converted

    try:
        x = float(x)
    except:
        warnings.warn(f"cant turn this value to numeric: {x}")
        return -1
        
    if x <= 0 or x >= 400000:
        return -1
    else:
        return x
```

```python
num_mileage = df['mileage'].dropna().apply(lambda x: to_plausible_mileage(x))
num_mileage = num_mileage[num_mileage > -1]
print(f"mileage: min: {num_mileage.min()} , mode: {num_mileage.mode().squeeze()}, max: {num_mileage.max()}")
min_mileage, max_mileage = num_mileage.min(), num_mileage.max()
```

```python
def clean_bool(x):
    if not isinstance(x, str):
        return False
    if "clean" in x:
        return True
    return False

df['title_status'].apply(lambda x: clean_bool(x)).value_counts()
```

## transmission

```python
def sort_transmission_type(x):
    if not isinstance(x, str):
        return "Unknown"
    if "automatic" in x.lower() or "cvt" in x.lower():
        return "Automatic"
    if "manual" in x.lower():
        return "Manual"
    if "smg" in x.lower() or "sequential" in x.lower():
        return "Sequential"
    return "Unknown"
df['transmission'].apply(lambda x: sort_transmission_type(x)).value_counts()
```

# Quick Clean up

```python
# cleaning
df['price'] = df['price'].apply(lambda x: to_plausible_price(x))
df['year'] = df['year'].apply(lambda x: to_plausible_year(x))
df['make'] = df['make'].replace("VW", "Volkswagen")
df['mileage'] = df['mileage'].apply(lambda x: to_plausible_mileage(x))
df['title_status'] = df['title_status'].apply(lambda x: clean_bool(x))
df['transmission'] = df['transmission'].apply(lambda x: sort_transmission_type(x))

# filtering
df = df[(df['price']>= min_price) & (df['price'] <= max_price)]
df = df[(df['year'] >= min_year) & (df['year'] <= max_year)]
df = df[(df['mileage'] >= min_mileage) & (df['mileage'] <= max_mileage)]
df = df.drop(columns=["Unnamed: 0"])
# these are too strict - maybe let autogluon encode them
# df = df[df['make'].isin(common_makes.index)]
# df = df[df['model_name'].isin(common_models.index)]
```

# Modeling with Autogluon

```python
from autogluon.tabular import TabularDataset, TabularPredictor
```

```python
ag_df = TabularDataset(df)
ag_df = ag_df.drop(columns=["uid"])
```

## eval via R^2 (biased by number of features)

```python
predictor = TabularPredictor(label="price", eval_metric="r2")
```

```python
predictor.fit(ag_df)
```

## Eval via RMSE , also train longer with expensive model zoo

```python
predictor2 = TabularPredictor(label="price", eval_metric="root_mean_squared_error")
predictor2.fit(ag_df,
    presets="best_quality",
    num_bag_folds=5, #5-fold CV
    )
```
